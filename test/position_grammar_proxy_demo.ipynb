# Region Position Grammar Demo

Visual check for FurnitureGroup-level position grammar evidence.

```python
from pathlib import Path
import sys

sys.path.insert(0, str(Path.cwd()))

from IPython.display import SVG, display

from features.scene_loader import load_structured_svg_scene
from features.scene_evaluator import evaluate_scene, scene_position_grammar_overlay_svg

def show_fixture(path):
    """Display region-level position grammar evidence for one SVG fixture."""
    scene = load_structured_svg_scene(path)
    matrix = evaluate_scene(scene)
    rows = {row["metric"]: row for row in matrix["rows"]}
    evidence = rows["P_position_fit"].get("region_position_grammar_evidence", [])
    print(
        path.name,
        "groups=", len(matrix.get("furniture_groups", [])),
        "position_items=", len(evidence),
        "fit=", rows["P_position_fit"]["score"],
        "confidence=", rows["P_position_confidence"]["score"],
    )
    for item in evidence:
        print(
            "  ",
            item["subject_id"],
            item["subject_kind"],
            item["position-grammar-type"],
            "members=", item["position-member-count"],
            "slots=", item["position-slot-count"],
            "residual_mean=", item["position-residual-mean"],
        )
    display(SVG(data=scene_position_grammar_overlay_svg(scene, matrix)))
```

```python
show_fixture(Path("test/SVG_baseline.svg"))
```
